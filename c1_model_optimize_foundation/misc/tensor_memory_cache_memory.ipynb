1) torch.cuda.memory_allocated(): The amount of GPU memory currently used by active tensors.
2) torch.cuda.memory_reserved(): The amount of GPU memory reserved by PyTorch's caching allocator.
   - [❌] It is not only cache memory. memory_reserved ≠ cache
   - [✅] It is total GPU memory held by PyTorch's caching allocator. memory_reserved = memory_allocated + cached_memory


```python
import torch
import gc

# Memory stats before deletion
print(" Step 1: Before Tensor Creation:")
print("Allocated(MB):", torch.cuda.memory_allocated()/(1024**2))
print("Reserved(MB) :", torch.cuda.memory_reserved()/(1024**2))

# Example: allocate a large tensor on GPU
x = torch.randn(10000, 10000, device='cuda')

# Memory stats before deletion
print("\n Step 2: After Tensor Creation:")
print("Allocated(MB):", torch.cuda.memory_allocated()/(1024**2))
print("Reserved(MB) :", torch.cuda.memory_reserved()/(1024**2))

# Step 1: Delete the tensor
del x

# Memory stats after deletion
print("\n Step 3: After tensor deletion , garbage collector NOT called, cache NOT cleared :")
print("Allocated(MB):", torch.cuda.memory_allocated()/(1024**2))
print("Reserved(MB) :", torch.cuda.memory_reserved()/(1024**2))

# Step 2: Force Python garbage collection
gc.collect()

print("\n Step 4: After tensor deletion , garbage collector called, cache NOT cleared :")
print("Allocated(MB):", torch.cuda.memory_allocated()/(1024**2))
print("Reserved(MB) :", torch.cuda.memory_reserved()/(1024**2))

# Step 3: Clear CUDA cache
torch.cuda.empty_cache()

print("\n Step 5: After tensor deletion , garbage collector called, cache cleared :")
print("Allocated(MB):", torch.cuda.memory_allocated()/(1024**2))
print("Reserved(MB) :", torch.cuda.memory_reserved()/(1024**2))
```
