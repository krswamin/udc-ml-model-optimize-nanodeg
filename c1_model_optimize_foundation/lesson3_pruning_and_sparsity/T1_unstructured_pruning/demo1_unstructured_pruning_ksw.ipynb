# Demo1: UNSTRUCTURED PRUNING: KSW Solution


### LLaMA 3.2 (3B) – Global Magnitude Pruning (20%) → Sparse CSR Export

This notebook:

1. Loads **TinyLlama/TinyLlama-1.1B-Chat-v1.0** in **FP32** (falls back to a public 1–3B-ish model if you don't have access).
2. Applies **20% global magnitude pruning** across all `nn.Linear.weight` tensors via `torch.nn.utils.prune.global_unstructured`.
3. **Removes** pruning reparameterizations so weights are *physically zeroed*.
4. Converts every linear **weight** to **CSR sparse** and saves a compact checkpoint.
5. Runs a tiny sanity generation before and after pruning.

````python
import torch, torch.nn as nn, math, gc

def gather_linear_weight_params(module):
    params = []
    for name, mod in module.named_modules():
        '''
        # The layer should be Linear and it should have a 'weight' attribute
        # then you just append it to a list of params like so
        [ (Linear(in_features=1024, out_features=512, bias=False), 'weight'), 
          (Linear(in_features=512, out_features=1024, bias=False), 'weight'),]
        '''       
        if isinstance(mod, nn.Linear) and getattr(mod, "weight", None) is not None:
            params.append((mod, "weight"))
    return params


'''
params_to_prune = 
[
(Linear(in_features=1024, out_features=512, bias=False), 'weight'), 
(Linear(in_features=512, out_features=1024, bias=False), 'weight'),
(Linear(in_features=1024, out_features=512, bias=False), 'weight'), 
(Linear(in_features=512, out_features=1024, bias=False), 'weight'),
]
'''
````

## Why Histogram Instead of Flattening
Flattening is the naive method
```
all_weights = torch.cat([W.flatten() for W in layers])
threshold = torch.quantile(all_weights.abs(), 0.2)
```

- For Large LLMs: 7B weights × 4 bytes = 28GB. You cannot allocate a 28GB vector
- 🎯 Instead you iterate layer by layer and build a histogram with 2048 bins. 2048 bins would take up ~16 KB. HUGE MEMORY SAVINGS 

````python

@torch.no_grad()
def _hist_threshold_for_global(cutoff_amount, params_to_prune, bins=2048):
    """
    First pass: build a global histogram of |weights| across all Linear layers (CPU),
    then find the magnitude threshold that prunes `cutoff_amount` fraction globally.
    """
    # 1) min/max over abs weights, streamed
    # gmin: smallest absolute weight seen so far
    # gmax: largest absolute weight seen so far
    # total_num_weights : total number of weights
    gmin, gmax = float("inf"), 0.0
    total_num_weights = 0
    for mod, pname in params_to_prune:
        # pname is "weight". so getattr(mod,weight) = mod.weight. It extracts the weight
        # detach()
        # tensors are usually part of a computation graph, used in gradient checking in the forward, backward pass
        # detach() creates a tensor view that shares memory with the original weight
        # but is disconnected from autograd. Operations on it do not track gradients.
        # The original weight tensor remains part of the computation graph.
        # IMPORTANT: detach() does NOT copy the data; it’s a view into the same memory.
        w = getattr(mod, pname).detach()
        # total number of weights
        total_num_weights += w.numel() 
        
        # move one layer at a time from gpu to cpu ( to prevent clogging the gpu)
        # calculate the statistics and then finally delete 'a' from gpuw = [-0.5, 0.3, -1.2]
        # w = [-0.5, 0.3, -1.2]
        # abs(w) = [0.5, 0.3, 1.2]: done on GPU
        # float(): covers abs(w) to fp32 from fp16, or fp8 or int8.: done on GPU
        # min, max are best calculated in fp32
        # then move w to CPU to calculate the min and max

        # Note : if you think GPU memory is tight already, move to cpu first and then do things like so
        # w_abs = w.cpu().abs().float()
        # nut note that GPU computation is very fast nevertheless
        w_abs = w.abs().float().cpu()
        gmin = min(gmin, float(w_abs.min()))
        gmax = max(gmax, float(w_abs.max()))
        del w_abs
        
    if not math.isfinite(gmin): gmin = 0.0
    if gmax <= gmin:  # all zeros?
        return 0.0, total_num_weights

    # 2) build histogram over [gmin, gmax]
    hist = torch.zeros(bins, dtype=torch.long)
    for mod, pname in params_to_prune:
        w = getattr(mod, pname).detach()
        w_abs = w.abs().float().cpu()
        
        # torch.histc returns float; use long counts
        # This counts the data in each bin. example
        # x = torch.tensor([0.1, 0.2, 0.5, 0.7]), torch.histc(x, bins=3, min=0.0, max=0.9)
        # Bin=0, Range = (0.0, 0.3), Count =2, 
        # Bin=1, Range = (0.3, 0.6), Count =1
        # Bin=2, Range = (0.6, 0.9), Count =1
        histogram_count = torch.histc(w_abs, bins=bins, min=gmin, max=gmax).to(torch.long)
        hist += histogram_count
        
        # in long loops , waiting for python garbage collection can cause temporary high peak memory usage
        # hence delete w,wabs,h. but this mostly optional
        del w
        del w_abs, histogram_count

    # if hist = torch.tensor([2, 5, 3, 1])
    # then the cumulative tensor is cum = tensor([2, 7, 10, 11])
    # For a cutoff, it is easier to work with cumulative histograme
    # in cum, 2 weights less than bin 1's lower limit. 7 weights less than bin 2's lower limit
    # 10 weights lower than nin 3's lower limit, 11 weights less than bin 4's lower limit
    cum = torch.cumsum(hist, dim=0)


    # 3) pick cutoff bin for desired prune count
    k = int(cutoff_amount * total_num_weights)
    # this is a safety clamp.
    # min(k, total_num_weights - 1), if cutoff_amount is 1 , it prevents accessing last index
    # max(0, ...) in case a negative cut_off amount was used
    # i.e k is the index and it has to be in the range [0,total_num_weights-1]
    k = max(0, min(k, total_num_weights - 1))
    # find the bin index in the cumulative histogram that corresponds to the k-th smallest weight.
    bin_idx = int(torch.searchsorted(cum, torch.tensor(k, dtype=torch.long)))
    bin_idx = max(0, min(bin_idx, bins - 1))


    # 4) convert bin index -> magnitude threshold (upper edge of that bin)
    bin_width = (gmax - gmin) / bins
    threshold = gmin + (bin_idx + 1) * bin_width
    
    return float(threshold), total_num_weights




@torch.no_grad()
def global_prune_linear_weights_streamed(model, amount=0.20, bins=2048):
    """
    Global unstructured magnitude pruning without allocating a giant vector on GPU.
    Finds a single global |w| threshold, then zeroes weights layer-by-layer.
    """
    params = gather_linear_weight_params(model)
    total = sum(getattr(m, p).numel() for m, p in params)
    print(f"Linear weights found: {len(params)} modules | {total/1e6:.2f}M params")

    threshold, total_num_weights = _hist_threshold_for_global(amount, params, bins=bins)
    print(f"Computed global threshold |w| < {threshold:.6g} to prune ~{amount:.0%} of {total_num_weights/1e6:.2f}M params")

    # Second pass: apply mask per layer (on current device)
    pruned = 0
    for mod, pname in params:
        W = getattr(mod, pname)
        mask = (W.abs() >= threshold)
        pruned += (mask.numel() - mask.sum()).item()
        W.mul_(mask.to(W.dtype))  # in-place zero
        del mask
    sparsity = pruned / max(1, total)
    print(f"Applied global pruning. Sparsity: {sparsity:.2%}")
    torch.cuda.empty_cache(); gc.collect()


    

````


## Convert Pruned Weights to CSR Tensor
- Original Weights: non sparse
- Pruned Weights: some percentage of values are made zero in the original weights. Hence this is a Sparse Matrix
- CSR Tensor: The Sparse matrix is converted to a tensor (there is an algorithm for this)
  NOTE: Only Linear Weights are Pruned and then converted to CSR. The Relu and other layers are as is99

### CSR TENSOR (COMPRESSED SPARSE ROW)
- This is simply for storing the weights. If you need to run inference with the Pruned weights, you'd still have to convert back to sparse matrix
- This is because nn.Linear still expects the weights in the original format
SPARSE INFERENCE: If CSR Tensors need to be used (instead of the pruned sparse matrix), the following things will be needed
- custom kernels
- torch.sparse.mm
- Triton kernels
- libraries like SparseGPT / DeepSparse





````python
# (Optional) Convert dense weights to sparse CSR tensors for storage/reporting.
# Note: PyTorch Linear expects dense weights for forward(). Use this only to export.
@torch.no_grad()
def export_sparse_state_dict(model):
    sd = {}
    for name, m in model.named_modules():
        if isinstance(m, nn.Linear) and getattr(m, "weight", None) is not None:
            W = m.weight.detach().to("cpu")
            # make CSR from 2D matrix
            idx = W.nonzero(as_tuple=False)
            if idx.numel() == 0:
                values = torch.tensor([], dtype=W.dtype)
                crow   = torch.zeros(W.size(0)+1, dtype=torch.int64)
                col    = torch.tensor([], dtype=torch.int64)
            else:
                # build CSR row pointers
                rows = idx[:, 0]
                cols = idx[:, 1]
                # sort by row then col
                order = torch.lexsort((cols, rows)) if hasattr(torch, "lexsort") else torch.argsort(rows*W.size(1)+cols)
                rows = rows[order]; cols = cols[order]
                values = W[rows, cols]
                crow = torch.zeros(W.size(0)+1, dtype=torch.int64)
                crow.index_add_(0, rows, torch.ones_like(rows, dtype=torch.int64))
                crow = torch.cumsum(crow, dim=0)
                col = cols.clone()
            sd[f"{name}.weight_csr"] = {
                "shape": tuple(W.shape),
                "crow_indices": crow,
                "col_indices": col,
                "values": values,
            }
    return sd
````

This exports a compact mapping `{linear_name.weight: CSR tensor}` to `pruned_sparse_ckpt/linear_weights_csr.pt`.

````python
from pathlib import Path

save_dir = Path("pruned_sparse_ckpt")
save_dir.mkdir(parents=True, exist_ok=True)

sparse_dump = {}
nonzero_total = 0
elem_total = 0

with torch.no_grad():
    for name, module in model.named_modules():
        if isinstance(module, nn.Linear) and hasattr(module, "weight") and module.weight is not None:
            W = module.weight.detach().to("cpu", copy=True)
            elem_total += W.numel()
            nonzero_total += (W != 0).sum().item()
            W_csr = W.to_sparse_csr()
            sparse_dump[f"{name}.weight"] = W_csr

torch.save(sparse_dump, save_dir / "linear_weights_csr.pt")
print("Saved CSR sparse weights to:", save_dir / "linear_weights_csr.pt")
print(f"Overall density: {nonzero_total/elem_total:.2%} | sparsity: {1 - nonzero_total/elem_total:.2%}")
````
